```python
# Definir pastas do projetos em variaveis para facilitar
bronze_path   = '/Volumes/bikestore/default/bikestore/bronze/'
silver_path   = '/Volumes/bikestore/default/bikestore/silver/'
gold_path     = '/Volumes/bikestore/default/bikestore/gold/'
resource_path = '/Volumes/bikestore/default/bikestore/resource/origem/'
#resource_path_volume = '/Volumes/infinitybigdata/azure/bikestore_resource/origem/'
```

```python
display(dbutils.fs.ls(resource_path))
```

```python
#Lendo arquivo
df_orders = spark.read.format('csv')\
.option('header','true')\
.option('inferSchema','true')\
.load(resource_path + 'orders.csv')
```

```python
display(df_orders)
```

```python
#Gravando o arquivo parquet como delta ao caminho
df_orders.write.mode('overwrite')\
.format('delta')\
.option('mergeSchema','true')\
.save(bronze_path + 'orders')
```

```python
#Verificando se o arquivo foi gravado
display(dbutils.fs.ls(bronze_path + 'orders'))
```
